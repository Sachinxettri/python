```python
students= [
    {"name":"ram","marks":90},
    {"name":"hari","marks":99},
    {"name":"sita","marks":68},
    {"name":"gita","marks":45},
]
marks =[student["marks"] for student in students]
print("Total student:",len(students))
print ("Heighst marks:",max(marks))
print ("lowest marks:",min(marks))
print ("Average mark:",sum(marks)/len(marks))
print("Results:")

for student in students:
    if student["marks"]>= 60:
        print(student["name"]," pass")
    else:
        print(student["name"]," fail")
```

```python
students= [
    {"name":"ram","marks":90},
    {"name":"hari","marks":99},
    {"name":"sita","marks":68},
    {"name":"gita","marks":59},
    {"name":"alina","marks":45}
]
passed = 0
failed = 0
for student in students:
    if student["marks"]>= 60:
        passed += 1
    else:
        failed += 1
topstudent = max (students, key=lambda s: s["marks"])
passpercantage = passed/len(students)*100
print("passed: " , passed)
print("failed: " , failed)
print("top student : ", topstudent["name"])
print ("pass percantage:",passpercantage)
        
#for student in students:
    #print(student["name"],student["grade"],student["marks"])
```
