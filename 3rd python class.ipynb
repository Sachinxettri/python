```python
students=[
    {"name":"sachin","grade":"A","marks":90},
    {"name":"ananta","grade":"C","marks":99},
{"name":"samir","grade":"B","marks":59},
{"name":"ram","grade":"D","marks":70}
]
```

```python
for student in students:
    print(student["name"],student["grade"],student["marks"])
```

```python
for student in students:
    if student["marks"]>= 60:
        print(student["name"],":pass") #bichama : yo rakna ko lagi chai pass ya fail ko agadi : yo rakda hunxa natra fail ko agadi bahira rakeko jastai ":" yo rakda ni hunca
    else:
        print(student["name"],":","fail")
        
```

```python
marks=[45,65,78,55,78]
passed=0
fail=0
for mark in marks:
    if mark>=60:
        passed +=1
    else:
        fail +=1
    print("passed"),passed)
    print("failed",fail)
```
