```python
data = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])
print("mean of all values:")
print(data.mean())

print("\nMean of each column:")
print(data.mean(axis=0))

print("\nMean of each row:")
print(data.mean(axis=1))
```
